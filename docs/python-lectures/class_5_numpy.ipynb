You can create a numpy array using `numpy.array(...)` function.
````python
powers = np.array([1, 2, 3])
add = np.arange(3, 6)
````

The main advantage of `numpy` is that it provides all the tools for operations vectorization.

````python
2**powers + add
````

The core of the package is written in C++. It means that a well optimized code with proper use of vectorized operations can result in substantial computation time gains. You should always try to utilize this feature as much as possible.

Just a short example for you:

````python
import time

def inner_product_list_implementation(a: np.ndarray, b: np.ndarray):  # list comprehension implementation of
    if a.size != b.size:                                              # inner product operation;
        raise ValueError(f"The two iterables must have the s")        # note that I will still use numpy arrays
    if a.ndim != 1 or b.ndim != 1:                                    # as arguments, but this won't help –
        raise NotImplemented()                                        # we still need a proper vectorization
    return [a * b for a, b in zip(a, b)]

def measure_time(func, *args, **kwargs):  # utility to measure computation time
    start = time.time()
    func(*args, **kwargs)
    end = time.time()
    print("--- time: %f ---" % (end - start))
    
    
a = np.random.normal(0, 1, 1000)
b = np.random.normal(0, 1, 1000)
print("For a simple implementation through a list comprehension:")
measure_time(inner_product_list_implementation, a, b)
print("For a numpy implementation:")
measure_time(np.inner, a, b)
# the time difference depends on the size of collections involved in computations
````

### Producing realizations of (pseudo-)random variables.

The module that is responsible for random variable simulations is called `numpy.random`. You can find some of the available distributions below.

````python
n_obs = 10

print(
      f"This is {n_obs} realizations of U[2, 5]:\n", 
      np.random.uniform(low=2, high=5, size=n_obs)
)

print(
      f"This is {n_obs} realizations of a r.v. uniformly distributed "
      "on integer values in [2, 5):\n", 
      np.random.randint(low=2, high=5, size=n_obs)
)

print(
      f"This is {n_obs} realizations of a gaussian r.v. with mean 2"
      "and st.dev. 4:\n", 
      np.random.normal(loc=2, scale=4, size=n_obs)
)
````

Note that these realizations are not purely random. In fact, nothing in your computer is purely random. All such procedures are often called "pseudo-random" because they always depend on something, e.g. a state of your hardware, software and other things.

So, to make your code reproducible, it is a good tone to specify a concrete value of a "seed" like this:
```
np.random.seed(1337)
```

### Linear algebra tools: `numpy.linalg`

The `linalg` module of numpy provides the standard linear-algebra operations: determinants, inverses, eigendecompositions, and solving linear systems. We will need one of these in a moment.

````python
sigma = np.ones((3, 3))
sigma[np.diag_indices_from(sigma)] = np.arange(1, 4)
print("We have the following matrix:\n", sigma)
print("Determinant:", np.linalg.det(sigma))
print("Inverse:\n", np.linalg.inv(sigma))
print("Eigenvalues and eigenvectors:", *np.linalg.eig(sigma), sep='\n')
print("Solve a simple linear system >> sigma * x = (0, 1, 2)^T:", np.linalg.solve(sigma, np.arange(3)), sep='\n')
````

A note on `np.linalg.inv` versus `np.linalg.solve`: if what you actually want is the solution of $A x = b$, use `solve`. Computing the inverse explicitly and then multiplying is both slower and less numerically accurate, and in practice you almost never need the inverse itself.

### Use case: linear regression

Imagine that you want to run a regression, namely estimate coefficients $b^*$ in the following data model:

$$\mathrm{Y} = \mathrm{X} b^* + \mathrm{W}$$

where $\mathrm{Y}$ is an outcome, $\mathrm{X}$ is the design matrix, and $\mathrm{W}$ is the noise.


Suppose for the moment that $\mathrm{Y}$ is one dimensional.

One way to look at the problem of estimating $b^*$ is to define a loss function, e.g. a squared error, and minimize it, getting an estimate: $$\mathrm{\hat{b}} \in \arg\min_{b} \, (\mathrm{Y} - \mathrm{X} \cdot b)^T (\mathrm{Y} - \mathrm{X} \cdot b)$$

First, let's generate some data from this model. Simulating our own data is useful precisely because we then *know* the true coefficients, so we can check how well any estimation procedure recovers them.

We use $d = 10$ covariates, of which only the first $s = 3$ are *active* (have a non-zero coefficient), and we let the covariates be correlated with each other with correlation $\rho$.

````python
np.random.seed(1337)

d = 10                   # number of covariates
s = 3                    # number of active covariates
rho = 0.5                # pairwise correlation between covariates

# covariance matrix with 1 on the diagonal and rho everywhere else
cov_X = (1 - rho) * np.eye(d) + rho

# true coefficients: 1, 1/2, 1/4 for the active ones, then zeros
b_active = (1/2)**np.arange(s)
b = np.concatenate((b_active, np.zeros(d - s)))
b
````

````python
n = 100

X = np.random.multivariate_normal(mean=np.zeros(d), cov=cov_X, size=n)
W = np.random.normal(size=n)
Y = X @ b + W

print("Shape of X:", X.shape)
print("Shape of Y:", Y.shape)
````

Now the estimation. Setting the gradient of the squared-error loss to zero gives the *normal equations*

$$\mathrm{X}^T \mathrm{X} \, \hat{b} = \mathrm{X}^T \mathrm{Y}$$

which is a linear system in $\hat{b}$ — exactly what `np.linalg.solve` exists for. Textbooks usually write the solution as $\hat{b} = (\mathrm{X}^T \mathrm{X})^{-1} \mathrm{X}^T \mathrm{Y}$, but as noted above, we solve the system rather than form the inverse.

Let's write a function that returns both the estimated coefficients and the *residual sum of squares*, $\mathrm{RSS} = (\mathrm{Y} - \mathrm{X}\hat{b})^T (\mathrm{Y} - \mathrm{X}\hat{b})$, which is the value of the loss at the minimum.

````python
def ols(X: np.ndarray, Y: np.ndarray) -> tuple[np.ndarray, float]:
    """
    Ordinary least squares.

    args:
        X:  design matrix, shape (n, d)
        Y:  outcome vector, shape (n,)

    returns:
        A tuple (b_hat, rss): the estimated coefficients, shape (d,),
        and the residual sum of squares.
    """
    b_hat = np.linalg.solve(X.T @ X, X.T @ Y)
    residuals = Y - X @ b_hat
    rss = residuals @ residuals          # same as np.sum(residuals**2)
    return b_hat, rss
````

````python
b_hat, rss = ols(X, Y)

print("True coefficients:     ", np.round(b, 3))
print("Estimated coefficients:", np.round(b_hat, 3))
print("Residual sum of squares:", round(rss, 3))
````

Two things to notice.

The active coefficients are recovered reasonably well, and the inactive ones come out close to zero but not exactly — with $n = 100$ noisy observations we should not expect an exact answer. Try increasing `n` and rerunning: the estimates should tighten around the truth.

$\mathrm{RSS} / n$ is roughly 1, which is the variance of the noise $\mathrm{W}$ we put in. That is not a coincidence — it is an estimate of exactly that quantity.

````python
# sanity check: the closed-form inverse and numpy's own least-squares routine agree
b_inv = np.linalg.inv(X.T @ X) @ X.T @ Y
b_lstsq = np.linalg.lstsq(X, Y, rcond=None)[0]

print("solve agrees with the inverse formula:", np.allclose(b_hat, b_inv))
print("solve agrees with np.linalg.lstsq:    ", np.allclose(b_hat, b_lstsq))
````

In practice you would reach for `np.linalg.lstsq` (or a library like `statsmodels` or `scikit-learn`) rather than write this yourself. The point here was to see that the whole procedure is a few lines of array arithmetic once you have `numpy`.
